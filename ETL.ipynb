# **TRANSFORMATION**

```python
# Drop rows where arr_flights is missing
df = df.dropna(subset=['arr_flights'])
```

```python
df.isnull().sum()
```

```python
missing_rows = df[df['arr_del15'].isnull()]
missing_rows
```

```python
# Fill missing arr_del15 with 0 because all other delay depended are 0 in arr_del15 rows i.e. no flight was delayed > 15 min...
df.loc[:, 'arr_del15'] = df['arr_del15'].fillna(0)
```

```python
df.isnull().sum()
```

```python
print("Dataset Shape:", df.shape)
```

```python
# Total delay per flight
df.loc[:,'total_delay'] = (
    df['carrier_delay'] +
    df['weather_delay'] +
    df['nas_delay'] +
    df['security_delay'] +
    df['late_aircraft_delay']
)

# Proportion of each delay cause
delay_cols = ['carrier_delay','weather_delay','nas_delay','security_delay','late_aircraft_delay']
for col in delay_cols:
    pct_col = col + '_pct'
    df.loc[:,pct_col] = df[col] / df['total_delay']



# Fill 0 for proportion rows where total_delay = 0
pct_cols = [col + '_pct' for col in delay_cols]
df.loc[:, pct_cols] = df[pct_cols].fillna(0)


# Time features
# Create 'quarter' column based on month (Northen Hemisphere)
# Q1: Jan–Mar → Winter
# Q2: Apr–Jun → Spring
# Q3: Jul–Sep → Summer
# Q4: Oct–Dec → Fall / Autumn

df['month'] = df['month'].astype(int)
df['quarter'] = ((df['month'] - 1) // 3 + 1)
```

```python
# Aggregated features (carrier/airport)
carrier_summary = df.groupby('carrier')['total_delay'].mean().reset_index().rename(columns={'total_delay':'avg_total_delay_per_carrier'})
airport_summary = df.groupby('airport')['total_delay'].mean().reset_index().rename(columns={'total_delay':'avg_total_delay_per_airport'})
```

```python
carrier_summary_sorted = carrier_summary.sort_values(
    by='avg_total_delay_per_carrier',
    ascending=False
)
carrier_summary_sorted.head(10)
```

```python
airport_summary_sorted = airport_summary.sort_values(
    by='avg_total_delay_per_airport',
    ascending=False
)
airport_summary_sorted.head(10)
```
